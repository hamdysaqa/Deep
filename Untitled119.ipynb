```python
import numpy as np
def stable_softmax(matrix):
    shifted = matrix - np.max(matrix, axis=1, keepdims=True)
    exp_values = np.exp(shifted)
    return exp_values / np.sum(exp_values, axis=1, keepdims=True)
def generate_embeddings(num_tokens, dim=4):
    np.random.seed(42)
    return np.random.rand(num_tokens, dim)
def initialize_weights(dim):
    return (
        np.random.rand(dim, dim),
        np.random.rand(dim, dim),
        np.random.rand(dim, dim)
    )
def self_attention_pipeline(text):
    words = text.split()
    embeddings = generate_embeddings(len(words))
    d_k = embeddings.shape[1]
    Wq, Wk, Wv = initialize_weights(d_k)
    queries = embeddings @ Wq
    keys = embeddings @ Wk
    values = embeddings @ Wv
    raw_scores = (queries @ keys.T) / np.sqrt(d_k)
    attention_probs = stable_softmax(raw_scores)
    context_vectors = attention_probs @ values
    return {
        "tokens": words,
        "scores": raw_scores,
        "weights": attention_probs,
        "output": context_vectors
    }
def main():
    user_input = input("Enter sentence: ").strip()
    if not user_input:
        print("No sentence provided!")
        return

    result = self_attention_pipeline(user_input)
    print("\nTOKENS:")
    print(result["tokens"])
    print("\nATTENTION SCORES:")
    print(result["scores"])
    print("\nATTENTION WEIGHTS:")
    print(result["weights"])
    print("\nOUTPUT (CONTEXT VECTORS):")
    print(result["output"])
if __name__ == "__main__":
    main()
```
